```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
@tool 
def check_prime_or_not(num:int):
    "check number is prime number or not"
    c=0
    for i in range(1,num+1):
        if num%i==0:
            c+=1
    
    if c==2:
        return('prime number')
    else:
        return('not prime number')
    
@tool
def check_perfect_or_not(num:int):
    "check number is perfect number or not"
    temp=num
    sum=0
    for i in range(1,num):
        if i%num==0:
            sum+=1
    if num==sum:
        return('perfect number')
    else:
        return('not perfect number')
    
tools=[check_perfect_or_not,check_prime_or_not]
```

```python
llm=ChatOllama(model="qwen2.5:1.5b",temperature=0)
```

```python
agent=create_agent(llm,tools)
```

```python
response=agent.invoke({
    'messages':[('user',"find given number is 7 prime or not?")]
})
# print(response)?
print(response['messages'][-1].content)
```

```python
response=agent.invoke({
    'messages':[('user',"find given number is 7 perfect or not?")]
})
# print(response)?
print(response['messages'][-1].content)
```
