```python
from langchain_google_genai import ChatGoogleGenerativeAI
import os
from dotenv import load_dotenv
```

```python
llm_model=ChatGoogleGenerativeAI(
    api_key=os.getenv("Gemini-Api"),
    model="gemini-3.5-flash"
)
```

```python
Response=llm_model.invoke("Explain Machine Learning ?")
'''
How it work?
    
    llm_model.invoke("Explain Machine Learning)
    sends the request to gemini model
    request =>" Explain Machine Learning ?"
    response =>AImessage()
'''
```

```python
print(Response)
```

```python
print(Response.content[0]['text'])
```

```python
while True:
    user=input("user:")
    Response=llm_model.invoke(user)
    print(f"AI Response:{Response}")
    choice=input(" Type \nNo - exit \nAny char")
    if choice.lower()=="no":
        break
    
```
