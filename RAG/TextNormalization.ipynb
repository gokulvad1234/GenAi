```python
from langchain_community.document_loaders import TextLoader
```

```python
data=TextLoader("ragflow.txt").load()
```

```python
print(data)
```

```python
data=data[0].page_content
```

# Converting character to lower case 

```python
data=data.lower()
print(data)
```

# Removing special char

```python
import re
import string 
```

```python
Var=string.punctuation
```

```python
data=re.sub(f'[{Var}]',' ',data)
```

```python
data=re.sub(r'\W',' ',data)
print(data)
```

# Removing the Extraspace

```python
data=re.sub(r'\s{2,}',' ',data)
print(data)
```
