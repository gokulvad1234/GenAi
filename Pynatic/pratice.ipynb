```python
from pydantic import BaseModel,Field
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate,PromptTemplate
```

```python
with open('data.txt','r+') as file:
    data=file.read()
```

```python
template='''
you are the helpful assistant that can answer question about the product.
answer the question based on the product details provided.
product :{product}
question :{question}
'''
```

```python
prompt_template=PromptTemplate.from_template(template)
```

```python
llm=ChatOllama(model="qwen2.5:1.5b",temperature=0)
```

```python
class product(BaseModel):
      name:str =Field(description='name of the product')
      price:str =Field(description='price of the product')
      description:str =Field(description='description of the product')
      category:str =Field(description='category of the product')
      
qwen=llm.with_structured_output(product)    
    
```

```python
chain=prompt_template | qwen
```

```python
user_input=[
    {
        'product':data,
        'question':'what s the price of the iphone 15 pro max?'
    },
    {
        'product':data,
        'question':'what is the price of galaxy s24 ultra?'
    },
    {
        'product':data,
        'question':'what is the price of the redmi'
    },
    {   'product':data,
        'question':'what is the price of the onepluse'}
]
```

```python
# u_input={'product':data,'question':'what is the onepluse 12r price?'}
# result=chain.invoke(u_input)
# print(result)
```

```python
response=chain.batch(user_input)# Batch is used to given multiple query
```

```python
result=[]
for i in response:
    result.append(i.append(i.model_dump_json()))
print(result)
```
