###Pydantic – Definition:

Pydantic is a Python data validation library that uses Python type annotations to validate, parse, and structure data. In LLM applications, it is used to ensure that the LLM's output follows a predefined data structure with the correct fields and data types.

Why do we use Pydantic in LLM?

We use Pydantic with LLMs to convert unstructured model responses into structured data, validate the output, and ensure that the application receives data in the expected format.

Real-time example: In an AI Resume Analyzer, Pydantic ensures that the LLM returns the candidate's name, skills, experience, and ATS score in the correct format.

```python
from pydantic import BaseModel, Field
from langchain_ollama import ChatOllama
from  langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
import os
```

```python
with open('data.txt','r+') as file:
    data=file.read()
```

### LLM Model

```python
qwen_model=ChatOllama(model="qwen2.5:1.5b",temperature=0)

# gemini_model=ChatGoogleGenerativeAI(
#     api_key=os.environ("Gemini-Api"),
#     model="gemini-3.5-flash"
# )
```

```python
template='''
you are the helpful assistant that can answer question about the product.
answer the question based on the product details provided.
product :{product}
question :{question}
'''
```

```python
prompt_template=PromptTemplate.from_template(template)
```

```python
class Product(BaseModel):
    name:str =Field(description='name of the product')
    price:str =Field(description='price of the product')
    description:str =Field(description='description of the product')
    category:str =Field(description='category of the product')
# gemini_model.with_structured_output(Product)
qwen=qwen_model.with_structured_output(Product)
```

```python
chain=prompt_template | qwen #This is LCEL
```

```python
user_input=[
    {
        'product':data,
        'question':'what s the price of the iphone 15 pro max?'
    },
    {
        'product':data,
        'question':'what is the price of galaxy s24 ultra?'
    },
    {
        'product':data,
        'question':'what is the price of the redmi'
    },
    {'product':data,
     'question':'what is the price of the onepluse'}
]
```
