# WordNetLemmatizer

```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
# nltk.download('wordnet')--> wordnet for lemmatization
```

```python
tokens=['houses','running','dogs','better','cats','flying','geese','studies','wolves','children','feet']
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
for word in tokens:
    print(word,lemmatizer.lemmatize(word))#pos---part of speech
```

```python
# nltk.download('averaged_perceptron_tagger')
# nltk.download("averaged_perceptron_tagger_eng")
```

```python
token_with_pos=nltk.pos_tag(tokens)
token_with_pos
```

""" it is part of speech start,
N--'n'
V---'v'

```python
def part_speech(word):
    if word[0]=="N":
        return "n"
    elif word[0]=='V':
        return "v"
    elif word[0]=='A':
        return 'a'
    else:
        return "n"
    
print(part_speech('VVA'))
```

```python
for token,pos in token_with_pos:
    print(token,"--->",lemmatizer.lemmatize(token,part_speech(pos)))
```
