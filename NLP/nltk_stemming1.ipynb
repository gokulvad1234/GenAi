# PorterStemmer

```python
from nltk.stem import PorterStemmer #  PorterStemmer supports only single language thats call english
stemmer=PorterStemmer()
```

```python
words = [
    "running", "runs", "runner", "played", "playing", "plays", "happily", "happiness",
    "studies", "studying", "studied", "flies", "flying", "connected", "connection",
    "connections", "relational", "national", "denationalization", "organization",
    "organizing", "organizations", "caring", "carefully", "fairness", "generously",
    "generosity", "ponies", "caresses", "cats", "dogs", "cats", "agreed", "disabled",
    "matting", "mating", "meeting", "milling", "messing", "sensational", "traditional",
    "communicate", "communicated", "communicating", "communication", "adjustable",
    "adjustment", "replacement", "required", "university"
]

```

```python
for word in words:
    print(f'{word}----->{stemmer.stem(word)}')
```
