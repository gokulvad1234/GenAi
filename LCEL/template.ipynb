* Step 1 => Prompt Template 

```python
prompt_template=ChatPromptTemplate.from_messages([
    ('system',"detect the native language from the given context and return only detected language"),
    ('human','{prompt}' )
])
prompt_template
```

* step 2==LLM

```python
llm=ChatGoogleGenerativeAI(
      model="gemini-3.8-flash",
      api_key=os.environ["Gemini-Api"],
      temperature=0.2
)
# res=llm.invoke('explain machine learning')
# print(res)
```

```python
ollama=ChatOllama(model="qwen2.5:1.5b")
```

* Step 3 =>Parser

```python
parser=StrOutputParser()
```

```python
language_chain=prompt_template |llm | parser
response=language_chain.invoke({'prompt' :"machine learning oru easyana laguage "})
```
