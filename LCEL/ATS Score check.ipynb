```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import ChatOllama
from langchain_community.document_loaders  import PyPDFLoader
import os 
```

## Pdf Loader

```python
data=PyPDFLoader('Gokul_Resume_Final.pdf').load()[0].page_content
print(data)
```

## LCEL ---Langchain Expression Language

###  Step 1 => Prompt Template 

```python
prompt_template=ChatPromptTemplate.from_messages([
    ('system','Analyze this resume and give an ATS score. inside the resume you should take only skill and job description.only return ATS Score and  Do not provide any explanation.'),
    ('human','{prompt}')
]
)
prompt_template
```

## step 2 ==>> LLM

```python
ollama=ChatOllama(model="qwen2.5:1.5b",temperature=0)
```

## Step 3 =>Parser

```python
parser=StrOutputParser()
```

```python
langchain_chain=prompt_template | ollama | parser
result=langchain_chain.invoke({'prompt':data})
```

```python
print(result)
```
