```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import ChatOllama
```

## Pdf Loader

```python
data=PyPDFLoader('Gokul_Resume_Final.pdf').load()[0].page_content
print(data)
```

### Combining multiple Lcel

### --->prompttemplate

```python
prompt_template1=ChatPromptTemplate([
    ('system','Analyze this resume and give an ATS score. inside the resume you should take only skill and job description.only return  ATS Score and actual content ATS score Do not provide any explanation.'),
    ('human','{prompt}')
]
)
```

### ---> LLM Model

```python
llm=ChatOllama(model="qwen2.5:1.5b",temperature=0)
```

```python
parser=StrOutputParser()
```

```python
lang_chain1=prompt_template1 | llm | parser
response=lang_chain1.invoke({'prompt':data})
```

```python
print(response)
```

```python
prompt_template2=ChatPromptTemplate([('system',"'You are a resume analyzer. how to imporve my ATS score."),
    ('human','{prompt}')
    
])
```

```python
lang_chain2=prompt_template2 | llm | parser

```

```python
full_chain=lang_chain1 | lang_chain2
```

```python
result=full_chain.invoke({'prompt':data})
```

```python
print(result)
```
