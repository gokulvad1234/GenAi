```python

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_community.document_loaders import PyPDFLoader
from pypdf import PdfReader
import os
```

## PdfLoader

```python
pdf=PyPDFLoader('Gokul_Resume_Final.pdf').load()[0].page_content
print(pdf)
```

## LCEL ---Langchain Expression Language
LangChain Expression Language (LCEL) is a way to connect different LangChain components together using the | (pipe) operator to create an AI application or workflow.

###  Step 1 => Prompt Template--Creates the input instruction.

```python
prompt_template=ChatPromptTemplate.from_messages([
    ('system',"'You are a resume analyzer. Extract only the skills from the given resume. Do not provide any explanation."),
    ('human','{prompt}')
])
```

## step 2 ==>> LLM---Generates an AI response

```python
ollama=ChatOllama(model="qwen2.5:1.5b",temperature=0)
```

## Step 3 =>Parser---Converts the AI response into the String format.

```python
parser=StrOutputParser()
```

```python
resume_chain=prompt_template | ollama | parser
result=resume_chain.invoke({'prompt':pdf})
```

```python
print(result)
```
