```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import ChatOllama
```

### Combining the Multiple chain

```python
translate_prompt_template=ChatPromptTemplate([
    ('system',
     'you are a helpful assistant.'
     'detect the language from the given text and translate to english language'
     'return only detected language and translated text'),
    ('human','{text}')
])
```

```python
llm=ChatOllama(model="qwen2.5:1.5b")
```

```python
parser=StrOutputParser()
```

```python
translate_chain=translate_prompt_template | llm | parser
response=translate_chain.invoke({'text':'Machine learning la neraya algorithm iruku,adula logistic '})
```

```python
summary_prompt_template=ChatPromptTemplate.from_messages([
    ('system',
     'you are helpful assistant'
     'summarize the given text in detected language from the translated text'),
    ('human','{text}')
])
```

```python
summarize_chain=summary_prompt_template | llm |parser
# final_resopnse=summarize_chain.invoke('text':trans)
```

```python
full_chain=translate_chain | summarize_chain
result=full_chain.invoke({'text':'machine learning oru easyna subject idthula neriya subject iruku'})
```
