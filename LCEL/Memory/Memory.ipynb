```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables import RunnableWithMessageHistory
```

### LLM Model

```python
llm=ChatOllama(model="qwen2.5:1.5b",temperature=0)
```

### ChatPromptTemplate

```python
prompt=ChatPromptTemplate.from_messages([
    ('system','you are the helpful assistant.'),
    MessagesPlaceholder(variable_name='history'),
    #to insert the message inside prompt template
    ('human','{input}')
])
```

```python
chain=prompt | llm | StrOutputParser() #LCEL
memory={}

def get_session_history(session_id):
    if session_id not in memory:
        memory[session_id]=InMemoryChatMessageHistory()#python memory # InMemoryChatMessageHistory() it is create a empty message is history
    return memory[session_id] # it return history

chain_with_memory=RunnableWithMessageHistory(
    chain,
    get_session_history,#session_id
    input_messages_key='input',
    history_messages_key='history' #store[session_id]--it stores in history variable
)
user1=chain_with_memory.invoke(
    {'input':'my name is alice'},
    config={'configurable':{'session_id':'user1'}}) # config is a keyward aruments


user2=chain_with_memory.invoke(
    {'input':'what is my name'},
    config={'configurable':{'session_id':'user1'}})# config is a keyward aruments
print(user2)
```
